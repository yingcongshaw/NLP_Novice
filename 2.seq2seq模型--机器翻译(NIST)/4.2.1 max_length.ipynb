# 4.2.1句对的最大长度的约定

```python
import matplotlib.pyplot as plt

lines1 = open('cn-eng.txt', encoding='UTF-8').read().strip().split('\n')
pairs1 = [[s for s in l.split('\t')] for l in lines1]
l = [len(pair[1].split()) for pair in pairs1]
plt.hist(l,rwidth=0.90)
plt.show()
```

如图所示，大量数据都分布在1-30之间，40以上很少，可以忽悠。故约定句对的最大长度为40。
